# Evo Steering Playground (Text-to-Image)

Use this notebook to run and compare `base`, `diff`, `max`, and `evolution` modes for FK diffusion steering.

Recommended start for evolution mode:
- `svgd_step_size`: 0.05 to 0.2
- `num_particles`: 4 to 8
- `time_steps`: 40 to 60

```python
%%bash
cd Fk-Diffusion-Steering
git checkout evo
git pull origin evo
python text_to_image/evo_steering_playground.py \
  --model-name stable-diffusion-2-1 \
  --prompt "a cinematic photo of a corgi astronaut on mars, ultra detailed" \
  --modes base,diff,max,evolution \
  --num-particles 4 \
    --guidance-reward-fn Clip-Score \
  --time-steps 50 \
  --lmbda 2.0 \
  --svgd-step-size 0.12 \
  --auto-fix-deps
```

```python
%%bash
cd Fk-Diffusion-Steering
git checkout evo
git pull origin evo

python text_to_image/evo_steering_playground.py \
  --model-name stable-diffusion-2-1 \
  --prompt "a cinematic photo of a corgi astronaut on mars, ultra detailed" \
  --modes base,diff,max,evolution \
  --guidance-reward-fn Clip-Score \
  --num-particles 4 \
  --time-steps 50 \
  --lmbda 2.0 \
  --svgd-step-sizes 0.03,0.06,0.10,0.15 \
  --seeds 1,2,3,4,5 \
  --auto-fix-deps

echo "\nLatest sweep outputs:"
ls -td text_to_image/output/evo_script_* | head -n 1
latest_dir=$(ls -td text_to_image/output/evo_script_* | head -n 1)
echo "Detail CSV: ${latest_dir}/scores_detail.csv"
echo "Summary CSV: ${latest_dir}/scores_summary.csv"
echo "Top summary rows:"
head -n 12 "${latest_dir}/scores_summary.csv"
```
