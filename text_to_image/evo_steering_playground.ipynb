```python
from pathlib import Path
import csv

output_root = Path("Fk-Diffusion-Steering/text_to_image/output")
runs = sorted(
    [p for p in output_root.glob("evo_script_*") if p.is_dir()],
    key=lambda p: p.stat().st_mtime,
    reverse=True,
 )

if not runs:
    print(f"No run folder found under: {output_root}")
else:
    latest = runs[0]
    summary_csv = latest / "scores_summary.csv"
    detail_csv = latest / "scores_detail.csv"

    print(f"Latest run: {latest}")
    print(f"Summary CSV: {summary_csv}")
    print(f"Detail CSV:  {detail_csv}")

    def preview_csv(path: Path, top_n: int = 10):
        if not path.exists():
            print(f"\n{path.name} not found")
            return

        print(f"\nTop {top_n} rows from {path.name}:")
        with path.open("r", encoding="utf-8", newline="") as f:
            reader = csv.DictReader(f)
            rows = list(reader)

        if not rows:
            print("(empty)")
            return

        # Sort by best_mean when available, otherwise keep original order.
        if "best_mean" in rows[0]:
            def _to_float(x):
                try:
                    return float(x)
                except Exception:
                    return float("-inf")
            rows = sorted(rows, key=lambda r: _to_float(r.get("best_mean", "")), reverse=True)

        cols = list(rows[0].keys())
        print(" | ".join(cols))
        print("-+-".join("-" * len(c) for c in cols))
        for row in rows[:top_n]:
            print(" | ".join(str(row.get(c, "")) for c in cols))

    preview_csv(summary_csv, top_n=12)
    preview_csv(detail_csv, top_n=12)
```

```python
%%bash
cd Fk-Diffusion-Steering
git checkout evo
git pull origin evo

python text_to_image/evo_steering_playground.py \
  --model-name stable-diffusion-2-1 \
  --prompt "a cinematic photo of a corgi astronaut on mars, ultra detailed" \
  --modes evolution \
  --guidance-reward-fn Clip-Score \
  --num-particles 4 \
  --time-steps 50 \
  --lmbda 2.0 \
  --svgd-step-sizes 0.03,0.06,0.10,0.15 \
  --seeds 1,2,3 \
  --resample-strategies multinomial,systematic,stratified,residual,none \
  --auto-fix-deps

echo "\nLatest strategy sweep outputs:"
latest_dir=$(ls -td text_to_image/output/evo_script_* | head -n 1)
echo "Run dir: ${latest_dir}"
echo "Summary CSV: ${latest_dir}/scores_summary.csv"
echo "Detail CSV:  ${latest_dir}/scores_detail.csv"

echo "\nTop summary rows (best mean_of_mean first):"
head -n 20 "${latest_dir}/scores_summary.csv"
```
