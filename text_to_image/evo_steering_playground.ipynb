```python
from pathlib import Path
import csv

output_root = Path("Fk-Diffusion-Steering/text_to_image/output")
runs = sorted(
    [p for p in output_root.glob("evo_script_*") if p.is_dir()],
    key=lambda p: p.stat().st_mtime,
    reverse=True,
 )

if not runs:
    print(f"No run folder found under: {output_root}")
else:
    latest = runs[0]
    summary_csv = latest / "scores_summary.csv"
    detail_csv = latest / "scores_detail.csv"

    print(f"Latest run: {latest}")
    print(f"Summary CSV: {summary_csv}")
    print(f"Detail CSV:  {detail_csv}")

    def preview_csv(path: Path, top_n: int = 10):
        if not path.exists():
            print(f"\n{path.name} not found")
            return

        print(f"\nTop {top_n} rows from {path.name}:")
        with path.open("r", encoding="utf-8", newline="") as f:
            reader = csv.DictReader(f)
            rows = list(reader)

        if not rows:
            print("(empty)")
            return

        # Sort by best_mean when available, otherwise keep original order.
        if "best_mean" in rows[0]:
            def _to_float(x):
                try:
                    return float(x)
                except Exception:
                    return float("-inf")
            rows = sorted(rows, key=lambda r: _to_float(r.get("best_mean", "")), reverse=True)

        cols = list(rows[0].keys())
        print(" | ".join(cols))
        print("-+-".join("-" * len(c) for c in cols))
        for row in rows[:top_n]:
            print(" | ".join(str(row.get(c, "")) for c in cols))

    preview_csv(summary_csv, top_n=12)
    preview_csv(detail_csv, top_n=12)
```

```python
from pathlib import Path
from collections import defaultdict
import re
import matplotlib.pyplot as plt
from PIL import Image

output_root = Path("Fk-Diffusion-Steering/text_to_image/output")
runs = sorted(
    [p for p in output_root.glob("evo_script_*") if p.is_dir()],
    key=lambda p: p.stat().st_mtime,
    reverse=True,
 )

if not runs:
    print(f"No run folder found under: {output_root}")
else:
    latest = runs[0]
    print(f"Using run: {latest}")

    # Match files like: diff_rs-residual_seed1_step0.1000.png
    pattern = re.compile(r"^(diff|evolution)_rs-residual_seed(\d+)_step([0-9.]+)\.png$")
    pairs = defaultdict(dict)

    for p in latest.glob("*.png"):
        m = pattern.match(p.name)
        if not m:
            continue
        mode, seed, step = m.group(1), int(m.group(2)), float(m.group(3))
        pairs[(seed, step)][mode] = p

    compare_keys = sorted(
        [k for k, v in pairs.items() if "diff" in v and "evolution" in v],
        key=lambda x: (x[1], x[0]),
    )

    if not compare_keys:
        print("No diff/evolution residual image pairs found.")
    else:
        max_rows = min(6, len(compare_keys))
        fig, axes = plt.subplots(max_rows, 2, figsize=(10, 4 * max_rows), dpi=140)

        if max_rows == 1:
            axes = [axes]

        for row_idx, (seed, step) in enumerate(compare_keys[:max_rows]):
            diff_path = pairs[(seed, step)]["diff"]
            evo_path = pairs[(seed, step)]["evolution"]

            diff_im = Image.open(diff_path).convert("RGB")
            evo_im = Image.open(evo_path).convert("RGB")

            left_ax, right_ax = axes[row_idx]
            left_ax.imshow(diff_im)
            left_ax.set_title(f"diff | seed={seed} | step={step:.4f}")
            left_ax.axis("off")

            right_ax.imshow(evo_im)
            right_ax.set_title(f"evolution (residual) | seed={seed} | step={step:.4f}")
            right_ax.axis("off")

        fig.suptitle("Visual comparison: diff vs evolution (residual strategy)", fontsize=14)
        fig.tight_layout()
        plt.show()
```
